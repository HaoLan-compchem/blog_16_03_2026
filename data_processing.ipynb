```python
import pandas as pd
import numpy as np
import rdkit
from rdkit import Chem
```

```python
df = pd.read_csv('core_smarts.csv')
```

```python
df
```

```python
df.rename(columns={'First(First(Core))': 'Smarts'}, inplace=True)
```

```python
df
```

```python
mol_list = df['Smarts'].apply(lambda x: Chem.MolFromSmarts(x)).to_list()
```

```python
from rdkit.Chem import Draw
img = Draw.MolsToGridImage(mol_list[:16], molsPerRow=5, subImgSize=(150,150))
img
```
